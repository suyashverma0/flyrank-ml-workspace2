## 3. Success metric

*One metric you can defend. What number means 'good'?*

The primary success metric would be Precision@K.

It measures how many of the top K recommended content items are actually useful for review.

A higher Precision@K means the ranking is better at prioritizing relevant content.

The metric can only be calculated once a reliable relevance label is available.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
k = 10

print("Success Metric: Precision@K")
print("K:", k)
print("Evaluation requires a reliable relevance label.")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule may not capture the different factors affecting content performance.

Content performance can depend on multiple signals such as CTR, engagement rate, scroll rate, and average position.

These signals may interact differently across content types and clients.

Machine learning can help combine these signals into a priority score.

The output will be used for decision support, not automatic content removal.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
signals = [
    "ctr",
    "engagement_rate",
    "scroll_rate",
    "avg_position"
]

available_signals = [col for col in signals if col in df.columns]

print("Available Performance Signals:")
print(available_signals)
```
